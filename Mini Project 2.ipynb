```python

import random

def generate_maze(rows, cols):
    # Each cell has walls: top, right, bottom, left
    maze = [
        [[1, 1, 1, 1] for _ in range(cols)]
        for _ in range(rows)
    ]

    visited = set()
    stack = [(0, 0)]
    visited.add((0, 0))

    directions = [
        (-1, 0, 0, 2),  # Up
        (0, 1, 1, 3),   # Right
        (1, 0, 2, 0),   # Down
        (0, -1, 3, 1)   # Left
    ]

    while stack:
        r, c = stack[-1]
        neighbors = []

        for dr, dc, wall, opposite in directions:
            nr, nc = r + dr, c + dc

            if (0 <= nr < rows and
                0 <= nc < cols and
                (nr, nc) not in visited):
                neighbors.append((nr, nc, wall, opposite))

        if neighbors:
            nr, nc, wall, opposite = random.choice(neighbors)

            # Remove wall between current and next cell
            maze[r][c][wall] = 0
            maze[nr][nc][opposite] = 0

            visited.add((nr, nc))
            stack.append((nr, nc))
        else:
            stack.pop()

    return maze


def display_maze(maze):
    rows = len(maze)
    cols = len(maze[0])

    for r in range(rows):
        # Print top walls
        line = ""
        for c in range(cols):
            line += "+"
            line += "---" if maze[r][c][0] else "   "
        line += "+"
        print(line)

        # Print left/right walls
        line = ""
        for c in range(cols):
            line += "|" if maze[r][c][3] else " "
            line += "   "
        line += "|" if maze[r][cols - 1][1] else " "
        print(line)

    # Print bottom boundary
    print("+" + "---+" * cols)


# Main program
rows = 10
cols = 10

maze = generate_maze(rows, cols)

print("MAZE GENERATED USING DFS")
display_maze(maze)
```
